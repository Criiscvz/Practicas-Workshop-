## 3. Análisis de Variables

### 3.1. Análisis de Variables Categóricas - Categóricas

#### **Sex - Survived:**

Las características a analizar, son:

* **Sex:** Sexo del pasajero.
* **Survived:** Indica si el pasajero sobrevivió. 0 para no y 1 para sí.


```python
# Muestre las primeras cinco filas del dataframe con df. Solo las columnas "Sex" y "Survived"
df[["Sex", "Survived"]].head(5)
```

Utilice la función [groupby](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html#pandas.DataFrame.groupby) para agrupar por un conjunto de características.

```python
# Agrupe los datos del dataframe df para las características 'Sex' y 'Survived'.
# Asigne el resultado en el dataframe dfGroup01
dfGroup01 = df.groupby(['Sex', 'Survived'])
```

```python
# Muestre el dataframe dfGroup01
dfGroup01
```

Utilice la función [count](https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.DataFrameGroupBy.count.html#pandas.core.groupby.DataFrameGroupBy.count) para contar los valores **en cada grupo**, excluyendo los valores faltantes.

```python
# Para cada uno de los grupos en el dataframe dfGroup01, cuente (función count) las observaciones en cada una de las características.
dfGroup01.count()
```

```python
# Para cada uno de los grupos en el dataframe dfGroup01, cuente (función count) las observaciones en la característica "Sex".
dfGroup01["Sex"].count()
```

```python
# Para cada uno de los grupos en el dataframe dfGroup01, cuente (función count) las observaciones en la característica "Survived".
dfGroup01["Survived"].count()
```

**¿Cómo interpretarías este resultado?**

**Interpretación:** Al agrupar los datos por el sexo del pasajero y ver si sobrevivieron o no, se nota una diferencia. En el caso de las mujeres, la gran mayoría logró salvarse: 233 sobrevivieron y 81 fallecieron. En cambio, con los hombres pasó todo lo contrario, ya que la gran mayoría no lo logró, 468 fallecieron y solo 109 se salvaron.

Esto demuestra el impacto que tuvo la regla de evacuación de "mujeres y niños primero" durante la tragedia. Las mujeres tuvieron una prioridad absoluta al momento de abordar los botes salvavidas, lo que explica por qué su tasa de supervivencia fue extremadamente alta en comparación con la de los hombres.

### Gráfico de barras por conteo


> Un caso concreto del gráfico de barras -y el más frecuente, por otro lado- es aquel en el que queremos que la longitud de la barra sea proporcional al número de elementos representado por ella. (*Count plot*, *2025*)

Utilice la función [countplot](https://seaborn.pydata.org/generated/seaborn.countplot.html) para contar de observaciones en cada contenedor categórico utilizando barras.

```python
# Muestre el gráfico de barras por conteo (función sns.countplot), con los atributos:
#   data con el valor df
#   x con el valor 'Sex'
sns.countplot(data=df, x="Sex")
```

**Interpretación:**
En este gráfico de barras (countplot) estamos viendo la cantidad total de pasajeros según su sexo. La barra de los hombres es muchísimo más alta, alcanzando un total de 577 pasajeros, mientras que la barra de las mujeres se queda bastante más abajo con 314 pasajeras.

Esto nos indica que la tripulación y los pasajeros a bordo del Titanic eran mayoritariamente hombres, superando casi por el doble a la población femenina disponible en el viaje.

```python
# Muestre el gráfico de barras por conteo (función sns.countplot) para el dataframe df.
# Utilice el atributo x con el valor 'Sex' y hue con el valor "Survived"
sns.countplot(data=df, x="Sex", hue="Survived")
```

**Interpretación:**
En este gráfico (countplot) estamos dividiendo el sexo de los pasajeros según si sobrevivieron (barra naranja, 1) o no (barra azul, 0). Se ve que para los hombres, la barra azul de fallecidos es grande comparada con la pequeña barra naranja de sobrevivientes. En cambio, para las mujeres la situación se invierte por completo.

Esto confirma de forma visual que el sexo fue un factor que aumentaba una probabilidad de vida o muerte en el Titanic.

### 3.2. Análisis de Variables Numéricas - Numéricas

#### **Survived - Age:**

Las características a analizar, son:

*   **Survived:** Indica si el pasajero sobrevivió. 0 para no y 1 para sí.
*   **Age:** Edad del pasajero en años.

```python
# Muestre las últimas cinco filas del dataframe con df. Solo las columnas "Survived" y "Age"
df[["Age","Survived"]].head(5)
```

Utilice la función [corr](https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.DataFrameGroupBy.corr.html#pandas.core.groupby.DataFrameGroupBy.corr) para calcular **fuerza** y **dirección** de la relación lineal entre dos variables numéricas, con valores entre **-1** y **1**.

```python
# Del dataframe df, solo para columnas "Survived" y "Age", muestre la correlación (función corr)
df[["Age","Survived"]].corr()
```

**¿Cómo interpretarías este resultado?**

**Interpretación:**
Al calcular la matriz de correlación lineal con la función corr() entre la edad y la supervivencia, vemos que el resultado es de -0.077221. Al ser un número tan cercano a 0, significa que prácticamente no existe una relación lineal directa entre la edad que tenía un pasajero y su probabilidad de salvarse.

El signo negativo (–) dice que a mayor edad, la probabilidad de sobrevivir disminuía un poco, pero el valor es tan bajo que no se puede tomar como una regla general. Esto ocurre porque, aunque se priorizó a los niños pequeños, dentro del grupo de los adultos la supervivencia dependió muchísimo más de otros factores más fuertes como el sexo de la persona o la clase en la que viajaba.

### Diagrama de dispersión

>  Es un diagrama de dispersión de las dos variables (*Diagrama de dispersión*, *2025*)


La función [relplot](https://seaborn.pydata.org/generated/seaborn.relplot.html#seaborn.relplot) proporciona acceso a varias funciones diferentes a nivel de eje que muestran la relación entre dos variables con asignaciones semánticas de subconjuntos.

```python
# Muestre un diagrama de dispersión (función sns.relplot), con los atributos:
#   data con el valor df
#   x con el valor 'Survived'
#   y con el valor 'Age'
sns.relplot(data=df,x="Survived",y="Age")
```

```python
# Muestre un diagrama de dispersión con un modelo de regresión lineal (función sns.regplot), con los atributos:
#   data con el valor df
#   x con el valor 'Survived'
#   y con el valor 'Age'
sns.regplot(data=df,x="Survived",y="Age")
```

**Interpretación:**
En estos gráficos (relplot y regplot) estamos cruzando una variable categórica como la supervivencia (0 o 1) con una variable numérica como la edad. Al graficar la dispersión, vemos dos columnas de puntos casi idénticas que van desde niños de 0 años es decir de meses, hasta adultos mayores, distribuidos de forma muy similar en el grupo de los que fallecieron (0.0) y en el de los que sobrevivieron (1.0).

Al trazar la línea de regresión, se nota que es prácticamente horizontal, mostrando una pendiente negativa ligera. Esto nos confirma que no hay un patrón claro entre la edad y la supervivencia. Salvo por el hecho de que casi todas las personas mayores de 65 años fallecieron, con la excepción de un único pasajero de 80 años que logró salvarse, la edad por sí sola no sirve para predecir si alguien iba a sobrevivir sin tomar en cuenta el sexo o la clase.

### 3.3. Análisis de Variables Categóricas - Numéricas

#### **Pclass - Age:**

Las características a analizar, son:

*   **Pclass:** Una de las 3 clases de billete: Clase 1, Clase 2 y Clase 3.
*   **Age:** Edad del pasajero en años.

```python
# Muestre las primeras ocho filas del dataframe con df. Solo las columnas "Pclass" y "Age"
df[['Age', 'Pclass']].head(8)
```

```python
# En dfGroup02, asigne los grupos del dataframe df para la característica 'Pclass'.
dfGroup02 = df.groupby('Pclass')
dfGroup02
```

```python
# Para dfGroup02 en la columna "Age", muestre las estadísticas descriptivas.
dfGroup02['Age'].describe()
```

**¿Cómo interpretarías este resultado?**

**Interpretación:**
Al sacar el describe() de la edad según la clase, se ve que la edad promedio disminuye a medida que bajamos de categoría. En Primera Clase el promedio de edad es de 38.2 años y la mitad de las personas (mediana) tienen más de 37 años. En Segunda Clase el promedio baja a 29.8 años, y en Tercera Clase el promedio es el más joven de todos, con 25.1 años y una mediana de 24 años.

Esto dice que las personas con tickets más costosos eran personas mayores de 30 años y los puestos más económicos los ocupaban personas más jóvenes.

### Diagramas de caja por categoría

> El diagrama de caja, también conocido como diagrama de caja y bigote o box plot, resume la distribución de los datos mostrándonos la posición de sus cuartiles y de los valores atípicos (outliers) (*Box Plot*, *2025*).



Utilice la función [boxplot](https://seaborn.pydata.org/generated/seaborn.boxplot.html) para mostrar la distribución de datos cuantitativos de una manera que facilita las comparaciones entre variables o entre niveles de una variable categórica.

```python
# Muestre el diagrama de caja por categoría (función sns.boxplot), con los atributos:
#   data con el valor df
#   x con el valor 'Pclass', y
#   y con el valor "Age"
sns.boxplot(data=df, x='Pclass', y='Age')
```

**Interpretación:**
El gráfico muestra que la mediana de edad disminuye según la clase: la Primera Clase es la más vieja, la Segunda es intermedia y la Tercera es la más joven. Esto confirma que a mayor nivel socioeconómico, mayor era la edad de los pasajeros.

### Gráfico de enjambre

> Esta función sitúa la variable categórica en el eje x y la variable cuantitativa en el eje y, mostrando con un punto cada uno de los valores. Para poder distinguir los puntos en el caso de que estén muy próximos, por defecto se aplica un "temblor" (jitter) a los puntos para separarlos del eje vertical de forma aleatoria, para que no se solapen (*Swarm plot*, *2025*).



Utilice la función [swarmplot](https://seaborn.pydata.org/generated/seaborn.swarmplot.html) para dibujar un diagrama de dispersión categórico con puntos ajustados para que no se superpongan.

```python
# Muestre el gráfico de enjambre (función sns.swarmplot), con los atributos:
#   data con el valor df
#   x con el valor 'Pclass', y
#   y con el valor "Age"
sns.swarmplot(data=df, x='Pclass', y='Age')
```

**¿Cómo interpretarías este resultado?**

**Interpretación:**
Este gráfico muestra cada pasajero como un punto. Se ve que en la Primera Clase los puntos están más dispersos hacia arriba (gente mayor), mientras que en la Tercera Clase hay una masa enorme de puntos concentrada entre los 15 y 30 años. Esto confirma que la Tercera Clase estaba llena de personas jóvenes.

#### ¿Es posible agrupar las variables **Pclass** y **Age**?

```python
# Agrupe las columnas "Pclass" y "Age". Aplique la función count a los grupos.
df.groupby(["Pclass", 'Age'])['Age'].count()
```
